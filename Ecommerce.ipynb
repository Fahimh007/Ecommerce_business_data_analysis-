```python
#Remove Duplicate Records
df.drop_duplicates(inplace=True)
```

```python
df.replace(["N/A","NULL"," "],np.nan,inplace=True)
```

```python
#Remove Extra Spaces
df=df.apply(lambda x:x.str.strip() if x.dtype=="object" else x)
```

```python
#Proper Case
df["Customer_Name"]=df["Customer_Name"].str.title()

df["City"]=df["City"].str.title()

df["State"]=df["State"].str.title()
```

```python
#fix eamils
df=df[df["Email"].str.contains("@",na=False)]
```

```python
#Convert Date 
df["Order_Date"]=pd.to_datetime(df["Order_Date"],errors="coerce")

df["Delivery_Date"]=pd.to_datetime(df["Delivery_Date"],errors="coerce")
```

```python
#Convert Numeric Columns
cols=["Qty","Unit_Price","Discount"]

for c in cols:
    df[c]=pd.to_numeric(df[c],errors="coerce")
```

```python
#Remove Negative Quantity
df=df[df["Qty"]>0]
```

```python
#fill missing values
df["Discount"].fillna(0,inplace=True)

df["Phone"].fillna("Unknown",inplace=True)

df["Delivery_Date"].fillna(df["Order_Date"],inplace=True)
```

```python
# create new columns
# Sales
df["Sales"]=df["Qty"]*df["Unit_Price"]
# Net Amount
df["Net_Amount"]=df["Sales"]-(df["Sales"]*df["Discount"]/100)
# Profit (Example)
df["Profit"]=df["Net_Amount"]*0.20
# Order Month
df["Month"]=df["Order_Date"].dt.month_name()
# Order Year
df["Year"]=df["Order_Date"].dt.year
# Order Weekday
df["Weekday"]=df["Order_Date"].dt.day_name()

```

```python
#Final checks
df.info()
```

```python
# Export Clean Data
df.to_excel("Clean_Ecommerce.xlsx",index=False)
# or
df.to_csv("Clean_Ecommerce.csv",index=False)
```
